```kotlin
import java.math.BigInteger

fun factorial(n: Int): Int {
    tailrec fun go(n: Int, acc: Int): Int =
        if (n <= 1)
            acc
        else
            go(n - 1, acc * n)
    return go(n, 1)
}
```

```kotlin
fun fib1(i: Int): Int =
    when {
        i == 0 -> 0
        i == 1 -> 1
        else -> fib(i - 1) + fib(i - 2)
    }

(0..50).map { fib1(it) }.joinToString()
```

```kotlin
fun fib(i: Int): Int {
    tailrec fun go(i1: Int, i2: Int, k: Int): Int =
        if (k == 0) {
            i1
        } else {
            go(i2, i1 + i2, k - 1)
        }
    return go(0, 1, i)
}

(0..50).map { fib(it) }.joinToString()
```

```kotlin
fun abs(n: Int): Int = if (n < 0) -n else n
```

```kotlin
fun formatResult(name: String, n: Int, f: (Int) -> Int): String =
    "The %s of %d is %d.".format(name, n, f(n))

println(formatResult("fatorial", 7, ::factorial))
println(formatResult("abs value", -42, ::abs))
```

```kotlin
fun <T> findFirst(ss: Array<T>, key: T): Int {
    fun loop(n: Int): Int =
        when {
            n >= ss.size -> -1
            ss[n] == key -> n
            else -> loop(n + 1)
        }
    return loop(0)
}

findFirst(arrayOf("a", "b"), "c")
```

```kotlin
fun <T> findFirst(ss: Array<T>, p: (T) -> Boolean): Int {
    fun loop(n: Int): Int =
        when {
            n >= ss.size -> -1
            p(ss[n]) -> n
            else -> loop(n + 1)
        }
    return loop(0)
}

findFirst(arrayOf("a", "b")) { it == "a" }
```

```kotlin
val <T> List<T>.tail: List<T>
    get() = drop(1)

val <T> List<T>.head: T
    get() = first()

tailrec fun <A> isSorted(aa: List<A>, order: (A, A) -> Boolean): Boolean =
    when {
        aa.isEmpty() || aa.tail.isEmpty() -> true
        order(aa.head, aa.tail.head) -> isSorted(aa.tail, order)
        else -> false
    }

isSorted(listOf(1, 2, 4, 10, 9)) { a1, a2 -> a1 < a2 }
isSorted(listOf(1)) { a1, a2 -> a1 < a2 }
isSorted<Int>(listOf()) { a1, a2 -> a1 < a2 }
```

```kotlin
fun <A, B, C> partial(a: A, f: (A, B) -> C): (B) -> C {
    val pf = fun (b: B): C {
        return f(a, b)
    }
    return pf
}

fun <A, B, C> partial(a: A, f: (A, B) -> C): (B) -> C =
    { b: B -> f(a, b) }
```

```kotlin
fun <A, B, C> curry(f: (A, B) -> C): (A) -> (B) -> C =
    { a: A -> { b: B -> f(a, b) } }
```

```kotlin
fun <A, B, C> uncurry(f: (A) -> (B) -> C): (A, B) -> C =
    { a: A, b: B -> f(a)(b) }
```

```kotlin
fun <A, B, C> compose(f: (B) -> C, g: (A) -> B) : (A) -> C =
    { a: A -> f(g(a)) }
```
